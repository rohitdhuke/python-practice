operator overloading 07/10/2026

```python
class Bankaccount:
    def __init__(self,name,balance):
        self.name=name
        self.balance=balance
    def __add__(self,a2):
        return self.balance + a2.balance
a1=Bankaccount('rohit',50000)
a2=Bankaccount('amit',50000)
print(a1+a2)
```

```python
class Product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def __gt__(self,p2):
        return self.price>p2.price
        
p1=Product('Loptop',68000)
p2=Product('printer',10200)
print(p1>p2)
```

```python
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    def __sub__(self,e2):
        return self.salary - e2.salary
e1 = Employee("Rohit", 60000)
e2 = Employee("Amit", 45000)

print(e1 - e2)
```
