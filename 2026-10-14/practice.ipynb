```python
from abc import ABC, abstractmethod
class A(ABC):
    @abstractmethod
    def show(self):
        pass

    @abstractmethod
    def display(self):
        pass

class B(A):
    def show(self):
        print("B show")

class C(A):
    def display(self):
        print("C display")
class D(B,C):
    pass
d=D()    
d.show()
d.display()
```

```python
from abc import ABC, abstractmethod

class A(ABC):
    @abstractmethod
    def show(self):
        pass

    def welcome(self):
        print("Welcome")

class B(A):
    def show(self):
        print("B show")

b = B()
b.welcome()   # Welcome: no need to override welcome()
```

```python
from abc import ABC, abstractmethod

class A(ABC):
    @abstractmethod
    def show(self):
        print("A abstract method body")

class B(A):
    def show(self):
        print("B show")

b = B()

b.show()      # B show
A.show(b)     # A abstract method body

# a = A()     # TypeError: creating an A object is blocked
```

Create:
- An abstract class Payment with abstract methods pay(amount) and refund(amount).
- A child class UPIPayment implementing both methods.
- Create its object and call both methods with amounts.

```python
from abc import ABC, abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay(self,amount):
        pass
    @abstractmethod
    def refund(self,amount):
       pass
class UPIPayment(Payment):
    def pay(self,amount):
        print(f"Paid ₹{amount}")

    def refund(self,amount):
        print(f"refund ₹{amount}")

obj=UPIPayment()
obj.pay(1000)
obj.refund(500)

```
