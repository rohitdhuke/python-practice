Assignment 1: Division Calculator
Problem Statement:
Write a Python program that accepts two numbers from the user and performs division using exception handling.
Requirements:
1. Accept two numbers using input().
2. Perform division (a / b).
3. Handle ZeroDivisionError if the denominator is zero.
4. Handle ValueError if the user enters invalid numeric input.
5. Use else to print the result when no exception occurs.
6. Use finally to print "Program completed".

```python
try:
    a=int(input('enter the first no'))
    b=int(input('enter the second no'))
    c= a / b
except ZeroDivisionError:
    print('division by 0 is not allowed')
except ValueError:
    print('invalid numeric input')
else:
    print(c)
finally:
    print('program completed')
```

1. Create a custom exception class InsufficientBalanceError that inherits from Exception.
2. Create a BankAccount class.
3. Use __init__(balance) to initialize the account balance.
4. Create a withdraw(amount) method.
5. Raise ValueError if the withdrawal amount is zero or negative.
6. Raise InsufficientBalanceError if the withdrawal amount exceeds the available balance.
7. Deduct the amount if the withdrawal is valid.
8. Create show_balance() to display the remaining balance.
9. Handle both exceptions outside the class using try-except.

```python
class InsufficientBalanceError(Exception):
    pass
class BankAccount:
    def __init__(self,balance):
        self.balance=balance
    def withdraw(self,amount):
        self.amount=amount
        if amount<=0:
            raise ValueError('invalid amount')
        elif amount > self.balance:
            raise InsufficientBalanceError('insufficient balance')
        else:
            self.balance-=amount
            print('withdrawal successfull',amount)
        
    def show_balance(self):
        print(self.balance)
obj=BankAccount(10000)
try:
    obj.withdraw(000)
    obj.show_balance()
except ValueError as e:
    print(e)

except InsufficientBalanceError as e:
    print(e)
   
    
    


        
```
