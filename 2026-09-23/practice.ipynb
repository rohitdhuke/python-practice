```python
# Given two integers M & N, calculate and return their multiplication using recursion without using  * operator
def cal_mul(m, n):

    # Base case
    if m == 0 or n == 0:
        return 0

    # Negative n
    if n < 0:
        a = cal_mul(m, n + 1)
        return a - m

    # Positive n
    a = cal_mul(m, n - 1)
    return m + a


m = int(input())
n = int(input())

print(cal_mul(m, n))
```

```python
def first_occurrence(arr, target):
    
    lower=0
    upper=len(arr)-1
    
    ans=-1
    while lower<=upper:
        mid=(lower+upper)//2
        if arr[mid]== target:
            ans=mid
            upper=mid -1            
        elif arr[mid]>target:
            upper=mid -1
        else:
            lower=mid +1
            
    return ans 


def last_occurrence(arr, target):
    # your code
    lower=0
    upper=len(arr)-1
    ans=-1
    while lower<=upper:
        mid=(lower+upper)//2
        if arr[mid]== target:
            ans=mid
            lower=mid + 1            
        elif arr[mid]>target:
            upper=mid -1
        else:
            lower=mid +1
            
    return ans 
    

def count_occurrence(arr, target):
    first= first_occurrence(arr, target)
    last= last_occurrence(arr, target)
    if first!=-1:
        count= last - first + 1
        return count
    else:
        return 0


arr = [1, 2, 2, 2, 2, 5, 7, 9, 12]
target = 2
print(count_occurrence(arr,target))
# Notes:
# Binary Search works on a sorted array and here we use the first & last target indexes.
# First occurrence → save mid and search LEFT.
# Last occurrence  → save mid and search RIGHT.
# Total count = last - first + 1.
# last - first gives the gaps/distance; +1 gives the total positions including both ends.
# ans = -1 means target is not found.

```

```python
def findfloor(arr,target):
    start=0
    end=len(arr)-1
    ans=-1
    while start<=end:
        mid=(start + end)//2
        if arr[mid]==target:
            return arr[mid]
        elif arr[mid]<target:
            ans=mid
            start=mid +1
        else:
            
            end=mid -1
    if ans!=-1:
        return arr[ans]
    else:
        return -1


arr = [2, 4, 7, 10, 13, 18]
target = 11
print(findfloor(arr,target))
```

```python
def missing_number(arr):
    i=0
    while i <= len(arr):
        for j in range(len(arr)):
            if i == arr[j]:
                break
                
        else:
            return i
        i+=1
    return arr
        


arr = [0, 1, 2, 4, 5]
print(missing_number(arr))
```

```python
# this is better approach(it checks during winding)
def find_target(arr,target):
    if len(arr)==0:
        return False
    if arr[0]==target:
        return True
    return find_target(arr[1:],target)
    
    
arr = [5, 8, 2, 10, 7]
target = 100
print(find_target(arr,target))
# Notes:
# Check arr[0] during winding.
# If target is found → return True immediately.
# Otherwise recursively check arr[1:].
# Empty array means target was not found → return False.
# No extra variable is needed because we directly return the recursive result.
```
