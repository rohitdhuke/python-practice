```python
# bubble sort in desending order
arr = [18, 7, 25, 4, 16, 10]
for i in range (len(arr)):
    for j in range(len(arr)-i-1):
        if arr[j]<arr[j+1]:
            arr[j],arr[j+1]=arr[j+1],arr[j]
print(arr)
```

```python
# Selection Sort in Descending Order
arr = [12, 5, 19, 3, 15, 8]

for i in range(len(arr)):
    max_idx=i
    for j in range (i+1,len(arr)):
        if arr[max_idx]<arr[j]:
            max_idx=j
    arr[i],arr[max_idx]=arr[max_idx],arr[i]
print(arr)
            
```

```python
# ascending order using Insertion Sort
arr = [11, 4, 8, 2, 9, 5]
for i in range(1,len(arr)):
    key=arr[i]
    j=i-1
    while j>=0 and arr[j]>key:
        arr[j+1]=arr[j]
        j-=1
    arr[j+1]=key
print(arr)

# key = arr[i]        → save current value
# j = i - 1           → start checking left side

# arr[j] > key
#     ↓
# shift arr[j] right
#     ↓
# j -= 1
#     ↓
# finally place key at arr[j + 1]



```

```python
#merge sort using recursion
def merge_sort(arr):
    while len(arr)<= 1:
        return arr

    mid=len(arr) // 2
    left=arr[:mid]
    right=arr[mid:]
    left=merge_sort(arr[:mid])
    right=merge_sort(arr[mid:])

    i=0 
    j=0
    result=[]
    while i < len(left) and j < len(right):
        if left[i]>right[j]:
            result.append(right[j])
            j+=1
        else:
            result.append(left[i])
            i+=1
    while i < len(left):
        result.append(left[i])
        i+=1
    while j < len(right):
        result.append(right[j])
        j+=1
    return result
        

arr = [14, 6, 9, 3, 11, 2, 8]
print(merge_sort(arr))
```

```python
#sort the array in ascending order by quick sort using recursion
def partition(arr,start,end):
    
    pivot=arr[start]
    i=start + 1
    count=0
    while i <= end:
        if arr[i] < pivot:
            count+=1
        i+=1
    new_pvt_idx = start + count
    arr[start],arr[new_pvt_idx]=arr[new_pvt_idx],arr[start]

    i=start
    j=end
    while i < new_pvt_idx  and j > new_pvt_idx :
        if arr[i] < pivot:
            i+=1
        elif arr[j] >= pivot:
            j-=1
            
        else:
            arr[i],arr[j]=arr[j],arr[i]
            i+=1
            j-=1
    return  new_pvt_idx

arr = [8, 3, 10, 2, 5, 7]
start = 0
end = 5
print(partition(arr,start,end))
```

```python
def partition(arr,start,end):
    pivot=arr[start]
    i=start+1
    count=0
    while i <= end:
        if arr[i] < pivot:
            count+=1
        i+=1
    pivot_index= start + count
    arr[start],arr[pivot_index]=arr[pivot_index],arr[start]
    i=start
    j=end
    while i < pivot_index and j > pivot_index:
        if arr[i]< pivot:
            i+=1
        elif arr[j]>=pivot:
            j-=1
        else:
            arr[i],arr[j]=arr[j],arr[i]
            i+=1
            j-=1
    return pivot_index
def quick_sort(arr,start,end):
    if start>=end:
        return arr
        
    pivot_idx = partition(arr,start,end)
    quick_sort(arr,start,pivot_idx - 1 )
    quick_sort(arr,pivot_idx + 1, end)
    return arr
        
        
arr = [12, 7, 14, 9, 3, 11, 6, 2]
print(quick_sort(arr,0,len(arr)-1))
```

```python
def binary_search(arr,target):
    low=0
    high=len(arr)-1
    while low<=high:
        mid=(low+high)//2
        if arr[mid]==target:
            return mid
        elif arr[mid]< target:
            low=mid+1
        elif arr[mid]>target:
            high=mid-1
    return -1


arr = [3, 7, 11, 18, 24, 31, 39, 45]
target = 31
print(binary_search(arr,target))
```
