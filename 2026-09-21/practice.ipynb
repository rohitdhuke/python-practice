```python
#sort array using selection sort in ascending order
arr = [64, 25, 12, 22, 11]
for i in range(len(arr)):
    min_index=i
    for j in range(i+1,len(arr)):
        if arr[j]<arr[min_index]:
            min_index=j
    arr[i],arr[min_index]=arr[min_index],arr[i]
print(arr)
```

```python
#sort array by ascending order using insertion sort
arr = [9, 5, 1, 4, 3]
for i in range(1,len(arr)):
    key=arr[i]
    j=i-1
    while j>=0 and arr[j]>key:
        arr[j+1]=arr[j]
        j-=1
    arr[j+1]=key
print(arr)
```

```python
#merge_sort ascending order
def merge_sort(arr):
    if len(arr)<=1:
        return arr
    mid=len(arr)//2
    left=arr[:mid]
    right=arr[mid:]
    left=merge_sort(left)
    right=merge_sort(right)
    i=0
    j=0
    result=[]
    while i < len(left) and j < len(right):
        if left[i]<right[j]:
            result.append(left[i])
            i+=1
        else:
            result.append(right[j])
            j+=1
    while i < len(left):
        result.append(left[i])
        i+=1
    while j < (len(right)):
        result.append(right[j])
        j+=1
    return result
    
arr = [12, 12, 5, 9, 3, 15, 6, 5]
print(merge_sort(arr))
```

```python
def partition(arr,start,end):
    pivot=arr[start]
    i=start+1
    count=0
    while i <= end:
        if arr[i]<pivot:
            count+=1
        i+=1
    new_index=start + count
    arr[start],arr[new_index]=arr[new_index],arr[start]
    i=start
    j=end
    while i < new_index  and j >= new_index :
        if arr[i] < pivot:
            i+=1
        elif arr[j]> pivot:
            j-=1  
        else:
            arr[i],arr[j]=arr[j],arr[i]
            i+=1
            j-=1
    return new_index
    
def quick_sort(arr,start,end):
    if start>=end:
        return arr

    pivot_index=partition(arr,start,end)
    quick_sort(arr,start,pivot_index-1)
    quick_sort(arr,pivot_index+1,end)
    return arr
        
    


    
arr = [7, 2, 9, 4, 3, 8, 5]    
start=0
end=len(arr)-1
print(quick_sort(arr,start,end))
```

```python
# tower of hanoi using recursion
def hanoi(n,source, helper,destination):
    if n==1:
        print(f' take disc from {source} and  insert into  -> {destination}')
        return

    hanoi(n-1,source,destination,helper)

    print(f'take disc from {source} and insert  into ->  {destination}')

    hanoi(n-1,helper,source,destination)

n=int(input('enter the number '))
hanoi(n,'a','b','c')
```
