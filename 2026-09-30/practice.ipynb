```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary

    def increase_salary(self, amount):
        # You write this part
        self.__salary+=amount

    def show_details(self):
        return self.__salary


e1 = Employee("Rohit", 50000)

e1.increase_salary(5000)
print(e1.show_details())
```

single inheritance: This is the simplest form of inheritance, where a class inherits from a
single base class


```python
class vechile:
    def move(self):
        print('car is moving')
class car(vechile):
    def start(start):
        print('car is started')
c1=car()
c1.start()
c1.move()
```

multiple inheritance

```python
class flyable:
    def fly(self):
        print('im flying')
class swimmable:
    def swim(self):
        print('im swmming')
class amphibian(flyable,swimmable):
    pass
amp=amphibian()
amp.fly()
amp.swim()
```

```python
class Father:
    def show(self):
        print("Father")


class Mother:
    def show(self):
        print("Mother")


class Child(Mother,Father):
    def show(self):
        print("Child")
        super().show()
c=Child()
c.show()
print(Child.mro())
```

```python
class a():
    pass
class b(a):
    pass
class c(b):
    pass
class d(c):
    pass
class e(d):
    pass
class f(e):
    pass
class g(f):
    pass
class h(g,f,e,c):
    pass
print(h.mro())
```
