```python
# Given two integers M & N, calculate and return their multiplication using recursion without using  * operator
def cal_mul(m, n):

    # Base case
    if m == 0 or n == 0:
        return 0

    # Negative n
    if n < 0:
        a = cal_mul(m, n + 1)
        return a - m

    # Positive n
    a = cal_mul(m, n - 1)
    return m + a


m = int(input())
n = int(input())

print(cal_mul(m, n))
```
