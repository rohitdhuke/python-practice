Vehicle should:
- Accept brand in __init__().
- Have a method show_details() that prints the brand.
Car should inherit Vehicle:
- Accept brand and model.
- Use super().__init__() to initialize the brand.
- Override show_details(), call super().show_details(), then print the model.
ElectricCar should inherit Car:
- Accept brand, model, and battery.
- Use super().__init__() appropriately.
- Override show_details(), call super().show_details(), then print the battery.

```python
class Vehicle:
    def __init__(self,brand):
        self.brand=brand
    def show_details(self):
        print('Brand:' ,self.brand)
class Car(Vehicle):
    def __init__(self,brand,model):
        self.model=model
        super().__init__(brand)
    def show_details(self):
        super().show_details()
        print('Model: ' ,self.model)
class ElectricCar(Car):
    def __init__(self,brand,model,battery):
        self.battery=battery
        super().__init__(brand,model)
    def show_details(self):
        super().show_details()
        print('battery:' ,self.battery)
e1 = ElectricCar("Tata", "Nexon EV", "45 kWh")
e1.show_details()
```

Problem Statement: Create three classes: A, B, and C.
- B inherits from A.
- C inherits from B.
- Each class has a show() method.
- Each show() prints its own class name.
- In B.show() and C.show(), use super().show() before printing their own class name.

```python
class A:
    def Show(self):
        print('A Class')
class B(A):
    def Show(self):
        super().Show()
        print('B Class')
class C(B,A):
    def Show(self):
        super().Show()
        print('C Class')
C1=C()
C1.Show()
```

Requirements:
- A has a show() method that prints "A Class".
- B overrides show(). First call super().show(), then print "B Class".
- C does the same: call super().show(), then print "C Class".
- D also overrides show(): call super().show(), then print "D Class".
- Create a D object and call show().
- Finally, print D.mro().

```python
class A:
    def Show(self):
        print('A Class')
class B(A):
    def Show(self):
        super().Show()
        print('B Class')
class C(A):
    def Show(self):
        super().Show()
        print('C Class')
class D(B,C):
    def Show(self):
        super().Show()
        print('D class')
d=D()
d.Show()
print(D.mro())
```

Create two completely separate classes
There should be no inheritance relationship between them.

```python
class Car:
    def start(self):
        print('car started')
class Bike:
    def start(self):
        print('bike started')
def start_vehicle(obj):
    obj.start()
c1=Car()
b1=Bike()
start_vehicle(c1)
start_vehicle(b1)
```
