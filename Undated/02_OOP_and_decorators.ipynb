```python
def login_required(func):

    def wrapper():
        # 1. Print "Checking login..."
        print('checking login')

        # 2. Execute the original function
        func()
    # 3. Return the wrapper function
    return wrapper

@login_required
def profile():
    print("Showing profile")


profile()
```

```python
logged_in = False

def login_required(func):

    def wrapper():
        if logged_in:
            func()
        else:
            print('pls login first')

        # If True → call original function

        # Otherwise → print "Please login first"

    return wrapper


@login_required
def profile():
    print("Showing profile")


profile()
```

```python
Create a decorator called show_details that:
1. Prints "Function started" before the original function.
2. Calls the original function with *args and **kwargs.
3. Stores the returned value in result.
4. Prints "Function finished" after the original function.
5. Returns result.
```

```python
def show_details(func):
    def wrapper(*args, **kwargs):
        print('function started')
        result=func(*args,**kwargs)
        
        print('function completed')
        
        return result
   
    return wrapper
@show_details
def org_func(a,b):
    return a * b

print(org_func(10,20))

```
