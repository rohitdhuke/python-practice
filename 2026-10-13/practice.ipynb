Create an abstract class named Employee containing an abstract method calculate_salary().
Create two child classes, FullTimeEmployee and PartTimeEmployee, that inherit from Employee.
Each child class must implement its own salary calculation logic.
Requirements
1. Import ABC and abstractmethod from the abc module.
2. Create an abstract class named Employee.
3. Define an abstract method named calculate_salary().
4. Create a FullTimeEmployee class that accepts name and monthly_salary.
5. Create a PartTimeEmployee class that accepts name, hours_worked, and hourly_rate.
6. Override calculate_salary() in both child classes.
7. For full-time employees, return the monthly salary.
8. For part-time employees, calculate salary as hours_worked * hourly_rate.

```python
from abc import ABC, abstractmethod
class Employee(ABC):
    @abstractmethod
    def calculate_salary(self):
        pass
class FullTimeEmployee(Employee):
    def __init__(self,name,monthly_salary):
        self.name=name
        self.monthly_salary=monthly_salary
    def calculate_salary(self):
        return self.monthly_salary
        
class PartTimeEmployee(Employee):
    def __init__(self,name,hours_worked,hourly_rate):
        self.name=name
        self.hours_worked=hours_worked
        self.hourly_rate=hourly_rate
    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate
e1 = FullTimeEmployee("Rohit", 50000)
e2 = PartTimeEmployee("Amit", 80, 500)

print(e1.calculate_salary())
print(e2.calculate_salary())
        
        
```

5. Override start() inside ElectricCar.
6. Create an object of ElectricCar and call start().

Create an abstract class Payment with an abstract method process_payment() that prints "Payment processing started".
Create a child class UPIPayment that inherits from Payment and overrides process_payment().
Inside the child method, use super() to execute the parent class's method first, then print "UPI payment successful".
Requirements
1. Create an abstract class Payment using ABC.
2. Declare process_payment() using @abstractmethod.
3. Include a print statement inside the abstract method.
4. Create UPIPayment and override process_payment().
5. Use super().process_payment() inside the child method.
6. Create a UPIPayment object and call process_payment().

```python
from abc import ABC, abstractmethod
class Payment(ABC):
    @abstractmethod
    def process_payment(self):
        print('payment processing started')
class UPIPayment(Payment):
    def process_payment(self):
        super().process_payment()
        print('UPI payment successful')
obj=UPIPayment()
obj.process_payment()

    
```

Create two abstract classes, Camera and MusicPlayer.
The Camera class must declare an abstract method take_photo(), and the MusicPlayer class must declare an abstract method play_music().
Create a class SmartPhone that inherits from both abstract classes and implements both methods.
Requirements
1. Use ABC and @abstractmethod.
2. Create Camera with the abstract method take_photo().
3. Create MusicPlayer with the abstract method play_music().
4. Create SmartPhone using multiple inheritance.
5. Implement both inherited abstract methods.
6. Create a SmartPhone object and call both methods.
7. o/p Photo captured successfully
Music is playing

```python
from abc import ABC, abstractmethod
class Camera(ABC):
    @abstractmethod
    def take_photo(self):
        pass
class MusicPlayer(ABC):
    @abstractmethod
    def play_music(self):
        pass
class SmartPhone(Camera, MusicPlayer):
    def take_photo(self):
        
        print('Photo captured successfully')
        

    def play_music(self):
        
        print('Music is playing')
        
obj=SmartPhone()
obj.take_photo()
obj.play_music()
```
