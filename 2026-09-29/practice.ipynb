__init__ → name, price
Static method → check_discount(price)
If price >= 1000, return "Discount Available"
Otherwise return "No Discount"
Create two objects:- "Laptop", 50000
- "Mouse", 500

Check both objects using the same static method.

```python
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    @staticmethod
    def check_discount(price):
        if price >= 1000:
            return 'discount available'
        else:
            return 'no discount'
p1=product('loptop',50000)
p2=product('mouse',500)
print(product.check_discount(p1.price))
print(product.check_discount(p2.price))
```

Create a class Employee:
- Class variable → company = "TCS"
- __init__ → name, salary
- Create a class method called change_company
- It should change company
- Change "TCS" to "Infosys"
- Print the new company name.

```python
class employee:
    company= 'tcs'
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    @classmethod
    def change_company(cls,new_company):
        cls.company=new_company
e1=employee('chris',50000)
print(e1.name,e1.salary,employee.company)
employee.change_company('infosys')
print(e1.name,e1.salary,employee.company)
```

Change the class-level company name.

If age is >= 18, return "Eligible", otherwise return "Not Eligible". This method only needs the value passed to it.

```python
class employee:
    company='tcs'
    def __init__(self,name,salary,age):
        self.name=name
        self.salary=salary
        self.age=age
    def increase_salary(self,amount):
        
        self.salary+=amount
    @classmethod
    def change_company(cls,new_company):
        cls.company= new_company
    @staticmethod
    def check_age(age):
        if age>=18:
            return 'eligible'
        else:
            return 'not eligible'
e1 = employee("Rohit", 50000, 25)
e2 = employee("Andrew", 40000, 17)
print(e1.name,e1.salary,e1.age,employee.company)
print(e2.name,e2.salary,e2.age,employee.company)
employee.change_company('infosys')
print('after changing company')
print(e1.name,e1.salary,e1.age,employee.company)
print(e2.name,e2.salary,e2.age,employee.company)
e1.increase_salary(5000)
print('after increasing salary of rohit ')
print(e1.name,e1.salary,e1.age,employee.company)
print('check age for both')
print(e1.check_age(e1.age))
print(e2.check_age(e2.age))

```

```python
class parentclass:
    # def __init__(self):
    #     self.name='parent class attribute'
    def parent_method(self):
        return 'this is parent method'
class childclass(parentclass):
    def __init__(self):
        super().__init__()
        self.child_name='child class attribute'
    def child_method(self):
        return 'this is child method'
obj=childclass()
# print(obj.name)
# print(obj.parent_method())
print(obj.child_name)
print(obj.child_method())
```

```python
def count_occurrence(arr, x):

    # Base case
    if len(arr) == 0:
        return 0

    # Check first element
    if arr[0] == x:
        # recursive call
        result=count_occurrence(arr[1:],x)
        return 1 + result
    else:
        # recursive call
        result=count_occurrence(arr[1:],x)
        return result


arr = [2, 5, 2, 7, 2, 8]
x = 2

print(count_occurrence(arr, x))
```

```python
def first_occur(arr, x):
    if len(arr) == 0:
        return -1

    elif arr[0] == x:
        return 0

    else:
        result = first_occur(arr[1:], x)

        if result == -1:
            return -1
        else:
            return result + 1


arr = [4, 7, 2, 7, 9, 7]
x = 7

print(first_occur(arr, x))
```
